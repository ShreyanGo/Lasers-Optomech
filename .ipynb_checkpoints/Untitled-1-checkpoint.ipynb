```python
# %% [markdown]
# # HW03 — Michelson Interferometers, Modulations, and Transfer Functions
# Syracuse University — Lasers and Optomechanics
#
# Works on paper. Let's see if it works in code.

# %%
%matplotlib widget
from ipywidgets import *

import sympy as sp
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

sp.init_printing(use_latex='mathjax')

# %%
plt.style.use('dark_background')

fontsize = 14
mpl.rcParams.update({
    "text.usetex": True,
    "figure.figsize": (9, 6),
    "figure.autolayout": True,
    "font.family": "serif",
    "font.serif": "georgia",
    "lines.linewidth": 1.5,
    "font.size": fontsize,
    "xtick.labelsize": fontsize,
    "ytick.labelsize": fontsize,
    "legend.fancybox": True,
    "legend.fontsize": fontsize,
    "legend.framealpha": 0.7,
    "legend.handletextpad": 0.5,
    "legend.labelspacing": 0.2,
    "legend.loc": "best",
    "axes.edgecolor": "#b0b0b0",
    "grid.color": "#707070",
    "xtick.color": "#b0b0b0",
    "ytick.color": "#b0b0b0",
    "savefig.dpi": 80,
    "pdf.compression": 9,
})

# %% [markdown]
# ---
# ## Part 1 — Homodyne Michelson
#
# Standard Michelson but with a local oscillator (LO) picked off from the input,
# phase-shifted by φ_hd, then beaten against the antisymmetric port field at a
# homodyne beamsplitter.
#
# The LO field is:
# $$E_\mathrm{LO} = -r_\mathrm{LO} e^{-i\phi_\mathrm{HD}} E_\mathrm{in}$$

# %% [markdown]
# ### 1.1 — Adjacency Matrix
#
# Write down every field in the system, then encode how each one depends on
# others in a matrix M. M[i,j] = contribution of field j to field i.
#
# Field vector (10 fields):
#
# | Symbol | What it is |
# |--------|-----------|
# | E1  | Input field |
# | E2  | Field hitting the Michelson BS |
# | E3  | X-arm, going toward end mirror |
# | E4  | Y-arm, going toward end mirror |
# | E5  | Returning from X-arm |
# | E6  | Returning from Y-arm |
# | E7  | Antisymmetric port field E_as |
# | E8  | Local oscillator E_lo |
# | E9  | Field at PD_A |
# | E10 | Field at PD_B |
#
# Optical elements:
# - M1: input BS (splits signal and LO), (r_m1, t_m1)
# - M2: homodyne steering mirror (imparts φ_hd), r_m2
# - M3: homodyne BS (beats E_as against E_lo), (r_hd, t_hd)

# %%
# define all symbolic variables
E1, E2, E3, E4, E5, E6, E7, E8, E9, E10 = sp.symbols('E1:11')

r_bs, t_bs = sp.symbols('r_bs t_bs', positive=True)
r_x,  r_y  = sp.symbols('r_x r_y',   positive=True)
r_m1, t_m1 = sp.symbols('r_m1 t_m1', positive=True)
r_m2       = sp.symbols('r_m2',       positive=True)
r_hd, t_hd = sp.symbols('r_hd t_hd', positive=True)

phi_x, phi_y, phi_hd = sp.symbols('phi_x phi_y phi_hd', real=True)

# %%
# build the 10x10 adjacency matrix entry by entry
M = sp.zeros(10, 10)

# %%
# E2 gets t_m1 * E1  (transmitted through input BS)
M[1, 0] = t_m1

# %%
# E3 = t_bs * e^{-i phi_x} * E2  (into X-arm)
M[2, 1] = t_bs * sp.exp(-sp.I * phi_x)

# %%
# E4 = -r_bs * e^{-i phi_y} * E2  (into Y-arm, reflection sign)
M[3, 1] = -r_bs * sp.exp(-sp.I * phi_y)

# %%
# E5 = -r_x * e^{-i phi_x} * E3  (reflected from X end mirror)
M[4, 2] = -r_x * sp.exp(-sp.I * phi_x)

# %%
# E6 = -r_y * e^{-i phi_y} * E4
M[5, 3] = -r_y * sp.exp(-sp.I * phi_y)

# %%
# E7 (AS port) = r_bs*E5 + t_bs*E6
M[6, 4] = r_bs
M[6, 5] = t_bs

# %%
# E8 (LO) = r_m1 * r_m2 * e^{-i phi_hd} * E1
M[7, 0] = r_m1 * r_m2 * sp.exp(-sp.I * phi_hd)

# %%
# E9 (PD_A) = t_hd*E7 + r_hd*E8
M[8, 6] = t_hd
M[8, 7] = r_hd

# %%
# E10 (PD_B) = -r_hd*E7 + t_hd*E8
M[9, 6] = -r_hd
M[9, 7] =  t_hd

# %%
# check it looks right
M

# %% [markdown]
# ### 1.2 — Electric Field Transfer Functions
#
# The system is E = M·E + source, which rearranges to (M - I)·E = -source.
# So invert (M - I) to get all transfer functions at once.
#
# We want:
# $$\frac{E_A}{E_\mathrm{in}} = G_{9,1}, \qquad \frac{E_B}{E_\mathrm{in}} = G_{10,1}$$

# %%
I10 = sp.eye(10)

# %%
# this takes a moment — sympy inverting a 10x10 symbolic matrix
G = (M - I10).inv()

# %%
# transfer function to PD_A
TF_A = sp.simplify(G[8, 0])
TF_A

# %%
# transfer function to PD_B
TF_B = sp.simplify(G[9, 0])
TF_B

# %% [markdown]
# ### 1.3 — Substitutions
#
# Plug in physically motivated simplifications:
# - Phase basis: φ_x = φ_c + φ_d, φ_y = φ_c - φ_d
# - Ideal 50:50 Michelson BS: r_bs = t_bs = 1/√2
# - Perfect end mirrors: r_x = r_y = 1
# - Perfect steering mirror: r_m2 = 1
# - Ideal 50:50 homodyne BS: r_hd = t_hd = 1/√2
# - Ideal 50:50 input BS: r_m1 = t_m1 = 1/√2

# %%
phi_c, phi_d = sp.symbols('phi_c phi_d', real=True)

# %%
subs = {
    phi_x : phi_c + phi_d,
    phi_y : phi_c - phi_d,
    r_bs  : 1/sp.sqrt(2),
    t_bs  : 1/sp.sqrt(2),
    r_x   : 1,
    r_y   : 1,
    r_m2  : 1,
    r_hd  : 1/sp.sqrt(2),
    t_hd  : 1/sp.sqrt(2),
    r_m1  : 1/sp.sqrt(2),
    t_m1  : 1/sp.sqrt(2),
}

# %%
EA = sp.simplify(TF_A.subs(subs))
EA

# %%
EB = sp.simplify(TF_B.subs(subs))
EB

# %% [markdown]
# Should get:
# $$\frac{E_A}{E_\mathrm{in}} = \frac{1}{4}\left(
#     -e^{-i2(\phi_c - \phi_d)} + e^{-i2(\phi_c+\phi_d)} - 2e^{-i\phi_{hd}}
# \right)$$
#
# $$\frac{E_B}{E_\mathrm{in}} = \frac{1}{4}\left(
#     e^{-i2(\phi_c - \phi_d)} - e^{-i2(\phi_c+\phi_d)} - 2e^{-i\phi_{hd}}
# \right)$$

# %% [markdown]
# ### 1.4 — Power Transfer Functions
#
# P = |E|^2 = E * conj(E).
# After factoring out e^{-i2φ_c}, combining exponentials into sin,
# then multiplying by the conjugate:
#
# $$P_A = \frac{1}{4}\left[1 + \sin^2(2\phi_d) + 2\sin(2\phi_d)\sin(2\phi_c - \phi_{hd})\right]$$
#
# $$P_B = \frac{1}{4}\left[1 + \sin^2(2\phi_d) - 2\sin(2\phi_d)\sin(2\phi_c - \phi_{hd})\right]$$

# %%
# compute symbolically
PA_sym = sp.simplify(EA * sp.conjugate(EA))
PA_sym

# %%
PB_sym = sp.simplify(EB * sp.conjugate(EB))
PB_sym

# %%
# sum and difference — these are what actually matter
Psum_sym = sp.simplify(PA_sym + PB_sym)
Psum_sym

# %%
Pdiff_sym = sp.simplify(PA_sym - PB_sym)
Pdiff_sym

# %% [markdown]
# P_diff is the error signal. Setting φ_hd = 2φ_c - π/2 maximises sensitivity to φ_d:
# $$P_\mathrm{diff}\big|_\mathrm{opt} = \sin(2\phi_d)$$

# %%
# numpy versions for plotting
def Pa(pc, pd, phd): return (3 - np.cos(4*pd) + 4*np.sin(2*pd)*np.sin(2*pc - phd)) / 8
def Pb(pc, pd, phd): return (3 - np.cos(4*pd) - 4*np.sin(2*pd)*np.sin(2*pc - phd)) / 8

# %%
phi_d_arr  = np.linspace(-np.pi, np.pi, 1000)
phi_c_arr  = np.linspace(-np.pi, np.pi, 1000)
phi_hd_arr = np.linspace(-np.pi, np.pi, 1000)

# %% [markdown]
# ### 1.5 — Interpretation
#
# #### Plot 1: Power vs φ_d
#
# Optimal homodyne angle φ_hd = 2φ_c - π/2, with φ_c = 0.
# P_A - P_B = sin(2φ_d) — crosses zero at φ_d = 0.
# That zero crossing is your dark-fringe lock point.

# %%
pc0     = 0.0
phd_opt = 2*pc0 - np.pi/2

# %%
fig, ax = plt.subplots()

pa_ = Pa(pc0, phi_d_arr, phd_opt)
pb_ = Pb(pc0, phi_d_arr, phd_opt)

ax.plot(phi_d_arr, pa_,       lw=2,          label=r'$P_A$')
ax.plot(phi_d_arr, pb_,       lw=2,          label=r'$P_B$')
ax.plot(phi_d_arr, pa_ + pb_, lw=2, ls='--', label=r'$P_A + P_B$')
ax.plot(phi_d_arr, pa_ - pb_, lw=2, ls='-.', label=r'$P_A - P_B$')

ax.axhline(0, color='w', lw=0.7)
ax.axvline(0, color='w', lw=0.7, ls=':')

ax.set_xlabel(r'Differential phase $\phi_d$ [rad]')
ax.set_ylabel('Normalised Power')
ax.set_title(r'Power vs $\phi_d$ — optimal $\phi_{hd}$, $\phi_c = 0$')
ax.set_xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
ax.set_xticklabels([r'$-\pi$', r'$-\pi/2$', r'$0$', r'$\pi/2$', r'$\pi$'])
ax.legend()
ax.grid()
plt.show()

# %% [markdown]
# #### Plot 2: Power vs φ_c
#
# φ_d = π/4 (bright fringe), φ_hd = 0.
# Common phase just shuffles power between detectors — sum stays flat.
# φ_c is a global phase, it doesn't change total output power.

# %%
phi_d_bright = np.pi / 4

# %%
fig, ax = plt.subplots()

pa2_ = Pa(phi_c_arr, phi_d_bright, 0.0)
pb2_ = Pb(phi_c_arr, phi_d_bright, 0.0)

ax.plot(phi_c_arr, pa2_,         lw=2,          label=r'$P_A$')
ax.plot(phi_c_arr, pb2_,         lw=2,          label=r'$P_B$')
ax.plot(phi_c_arr, pa2_ + pb2_,  lw=2, ls='--', label=r'$P_A + P_B$')
ax.plot(phi_c_arr, pa2_ - pb2_,  lw=2, ls='-.', label=r'$P_A - P_B$')

ax.axhline(0, color='w', lw=0.7)
ax.set_xlabel(r'Common phase $\phi_c$ [rad]')
ax.set_ylabel('Normalised Power')
ax.set_title(r'Power vs $\phi_c$ — $\phi_d = \pi/4$, $\phi_{hd} = 0$')
ax.set_xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
ax.set_xticklabels([r'$-\pi$', r'$-\pi/2$', r'$0$', r'$\pi/2$', r'$\pi$'])
ax.legend()
ax.grid()
plt.show()

# %% [markdown]
# #### Plot 3: Power vs φ_hd
#
# φ_d = π/4 (bright fringe), φ_c = 0.
# Sweeping the homodyne angle rotates the interference between E_as and E_lo.
#
# Catch: this only works on a bright fringe. If E_as = 0 (dark fringe),
# the LO has nothing to beat against and φ_hd does nothing.
# This is why real experiments need a coherent locking field (CLF).

# %%
fig, ax = plt.subplots()

pa3_ = Pa(0.0, phi_d_bright, phi_hd_arr)
pb3_ = Pb(0.0, phi_d_bright, phi_hd_arr)

ax.plot(phi_hd_arr, pa3_,         lw=2,          label=r'$P_A$')
ax.plot(phi_hd_arr, pb3_,         lw=2,          label=r'$P_B$')
ax.plot(phi_hd_arr, pa3_ + pb3_,  lw=2, ls='--', label=r'$P_A + P_B$')
ax.plot(phi_hd_arr, pa3_ - pb3_,  lw=2, ls='-.', label=r'$P_A - P_B$')

ax.axhline(0, color='w', lw=0.7)
ax.set_xlabel(r'Homodyne angle $\phi_{hd}$ [rad]')
ax.set_ylabel('Normalised Power')
ax.set_title(r'Power vs $\phi_{hd}$ — $\phi_d = \pi/4$, $\phi_c = 0$')
ax.set_xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
ax.set_xticklabels([r'$-\pi$', r'$-\pi/2$', r'$0$', r'$\pi/2$', r'$\pi$'])
ax.legend()
ax.grid()
plt.show()

# %% [markdown]
# ---
# ## Part 2 — Asymmetric Michelson
#
# Now L_x >> L_y. Sidebands generated by mirror motion pick up different
# phase delays in the two arms because k = Ω/c and each frequency component
# sees a different Ω. So the Michelson response becomes frequency-dependent.
#
# Goal: derive P_as / φ_CARM(ω) and P_as / φ_DARM(ω).

# %% [markdown]
# ### 2.1 — Field Transfer Functions
#
# Simple one-bounce Michelson. Write down E_x, E_y, E_as in terms of E_in.

# %%
# YOUR WORK HERE
# E_x / E_in =
# E_y / E_in =
# E_as / E_in =

# %% [markdown]
# ### 2.2 — End Mirror Modulation
#
# Apply common CARM modulation: φ_x(t) = φ_y(t) = Γ cos(ωt)
#
# Small modulation (Γ << 1), expand e^{-iΓcos(ωt)} to first order in Γ.
# You should get three frequency components: ω_0, ω_0+ω, ω_0-ω.

# %%
# YOUR WORK HERE
# E_x(t) / E_in =
# E_y(t) / E_in =

# %% [markdown]
# ### 2.3 — Propagate to the AS Port
#
# Each field propagates back with phase e^{-ikL} where k = Ω/c.
# The carrier, upper sideband, and lower sideband all have different Ω,
# so they accrue different phases on the way back.
#
# Carrier:        φ_x0 = ω_0 L_x / c
# Upper sideband: φ_x,usb = φ_x0 (1 + ω/ω_0)
# Lower sideband: φ_x,lsb = φ_x0 (1 - ω/ω_0)

# %%
# YOUR WORK HERE
# E_as(t) / E_in =

# %% [markdown]
# ### 2.4 — Power Response to CARM Motion
#
# P_as = |E_as|^2. Cross terms between carrier and sidebands give the
# time-varying signal. Keep only first order in Γ (drop Γ^2 ≈ 0).
#
# Answer (given):
# P_as/P_in = r_x^2 [ sin^2(2φ_d) - 2Γ sin(4φ_d) sin(ω φ_d/ω_0) sin(ωt + ω φ_c/ω_0) ]

# %%
# YOUR WORK HERE — derive the above using sympy if needed

# %% [markdown]
# ### 2.5 — Demodulation
#
# Demodulate by integrating P_as over one cycle of ωt, multiplied by:
# - 1          → DC term P_as^DC
# - cos(ωt)   → I quadrature P_as^I
# - sin(ωt)   → Q quadrature P_as^Q

# %%
# DC term
# P_as^DC =

# %%
# I quadrature
# P_as^I =

# %%
# Q quadrature
# P_as^Q =

# %% [markdown]
# ### 2.6 — CARM Frequency Response
#
# Full complex response: P_as(ω) = P_as^I + i P_as^Q
# Plot magnitude and phase of P_as/Γ(ω).
#
# Parameters:
# r_x = 1, L_x = 1000 m, L_y = 1 m, λ = 1064 nm

# %%
c      = 3e8
lam    = 1064e-9
omega0 = 2*np.pi * c / lam

# %%
Lx = 1000.0
Ly = 1.0
rx = 1.0

# %%
phi_x0_val = omega0 * Lx / c
phi_y0_val = omega0 * Ly / c

phi_c0_val = (phi_x0_val + phi_y0_val) / 2
phi_d0_val = (phi_x0_val - phi_y0_val) / 2

print(f"phi_x0 = {phi_x0_val:.4e} rad")
print(f"phi_y0 = {phi_y0_val:.4e} rad")
print(f"phi_c0 = {phi_c0_val:.4e} rad")
print(f"phi_d0 = {phi_d0_val:.4e} rad")

# %%
f_arr     = np.logspace(0, 6, 2000)
omega_arr = 2 * np.pi * f_arr

# %%
# YOUR WORK HERE — plug in your P_as^I and P_as^Q expressions and form the TF
# TF_CARM = P_as^I + 1j * P_as^Q

# %%
# plot magnitude
fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True, figsize=(9, 8))

# ax1.loglog(f_arr, np.abs(TF_CARM), ...)
ax1.set_ylabel(r'$|P_\mathrm{as}/\Gamma|$')
ax1.set_title(r'CARM Frequency Response — Asymmetric Michelson')
ax1.grid(which='both', alpha=0.4)

# %%
# plot phase
# ax2.semilogx(f_arr, np.angle(TF_CARM, deg=True), ...)
ax2.set_ylabel(r'Phase [deg]')
ax2.set_xlabel(r'Frequency [Hz]')
ax2.set_yticks([-180, -90, 0, 90, 180])
ax2.grid(which='both', alpha=0.4)

plt.tight_layout()
plt.show()
```
